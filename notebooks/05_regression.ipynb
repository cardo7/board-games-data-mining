# Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold, RepeatedStratifiedKFold
```

```python
df = pd.read_csv('../data/processed/board_games_clean.csv')
```

```python
df['NumWish'].corr(df['NumOwned'])
```

```python
df_train, df_test = train_test_split(df, test_size=0.3, random_state=100)
```

```python
x_train = df_train[['NumOwned']]
y_train = df_train['NumWish']

x_test = df_test[['NumOwned']]
y_test = df_test['NumWish']
```

## Linear

```python
linear = LinearRegression()
linear.fit(x_train, y_train)

ridge = Ridge()
ridge.fit(x_train, y_train)

lasso = Lasso()
lasso.fit(x_train, y_train)
```

```python
print('Generalization error\n')

print('LINEAR'.ljust(15), 'RIDGE'.ljust(15), 'LASSO'.ljust(15))
print('R2: %.3f'.ljust(14) % r2_score(y_test, linear.predict(x_test)),
      'R2: %.3f'.ljust(14) % r2_score(y_test, ridge.predict(x_test)),
      'R2: %.3f'.ljust(15) % r2_score(y_test, lasso.predict(x_test)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_test, linear.predict(x_test)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_test, ridge.predict(x_test)),
      'MSE: %.3f'.ljust(15) % mean_squared_error(y_test, lasso.predict(x_test)))
print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, linear.predict(x_test)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, ridge.predict(x_test)),
      'MAE: %.3f'.ljust(15) % mean_absolute_error(y_test, lasso.predict(x_test)))

print('\n')
print('Loss\n')

print('LINEAR'.ljust(15), 'RIDGE'.ljust(15), 'LASSO'.ljust(15))
print('R2: %.3f'.ljust(14) % r2_score(y_train, linear.predict(x_train)),
      'R2: %.3f'.ljust(14) % r2_score(y_train, ridge.predict(x_train)),
      'R2: %.3f'.ljust(15) % r2_score(y_train, lasso.predict(x_train)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_train, linear.predict(x_train)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_train, ridge.predict(x_train)),
      'MSE: %.3f'.ljust(15) % mean_squared_error(y_train, lasso.predict(x_train)))
print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, linear.predict(x_train)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, ridge.predict(x_train)),
      'MAE: %.3f'.ljust(15) % mean_absolute_error(y_train, lasso.predict(x_train)))
```

## Multivariate

```python
features = ['NumOwned', 'ComWeight', 'YearPublished']
target = 'NumWish'
```

```python
x = df[features]
y = df[target]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=100)
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
linear = LinearRegression()
linear.fit(x_train, y_train)

ridge = Ridge()
ridge_cv = RidgeCV(alphas=np.logspace(-3, 3, 100), cv=5)
ridge.fit(x_train, y_train)
ridge_cv.fit(x_train_scaled, y_train)

lasso = Lasso()
lasso_cv = LassoCV(alphas=np.logspace(-3, 3, 100), cv=5)
lasso.fit(x_train, y_train)
lasso_cv.fit(x_train_scaled, y_train)
```

```python
print('Generalization error\n')

print('LINEAR'.ljust(15), 'RIDGE'.ljust(15), 'RIDGE CV'.ljust(15), 'LASSO'.ljust(15), 'LASSO CV'.ljust(15))
print('R2: %.3f'.ljust(14) % r2_score(y_test, linear.predict(x_test)),
      'R2: %.3f'.ljust(14) % r2_score(y_test, ridge.predict(x_test)),
      'R2cv: %.3f'.ljust(15) % r2_score(y_test, ridge_cv.predict(x_test_scaled)),
      'R2: %.3f'.ljust(15) % r2_score(y_test, lasso.predict(x_test)),
      'R2cv: %.3f'.ljust(15) % r2_score(y_test, lasso_cv.predict(x_test_scaled)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_test, linear.predict(x_test)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_test, ridge.predict(x_test)),
      'MSEcv: %.3f'.ljust(15) % mean_squared_error(y_test, ridge_cv.predict(x_test_scaled)),
      'MSE: %.3f'.ljust(15) % mean_squared_error(y_test, lasso.predict(x_test)),
      'MSEcv: %.3f'.ljust(15) % mean_squared_error(y_test, lasso_cv.predict(x_test_scaled)))

print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, linear.predict(x_test)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, ridge.predict(x_test)),
      'MAEcv: %.3f'.ljust(15) % mean_absolute_error(y_test, ridge_cv.predict(x_test_scaled)),
      'MAE: %.3f'.ljust(15) % mean_absolute_error(y_test, lasso.predict(x_test)),
      'MAEcv: %.3f'.ljust(15) % mean_absolute_error(y_test, lasso_cv.predict(x_test_scaled)))

print('\n')
print('Loss\n')

print('LINEAR'.ljust(15), 'RIDGE'.ljust(15), 'RIDGE CV'.ljust(15), 'LASSO'.ljust(15), 'LASSO CV'.ljust(15))
print('R2: %.3f'.ljust(14) % r2_score(y_train, linear.predict(x_train)),
      'R2: %.3f'.ljust(14) % r2_score(y_train, ridge.predict(x_train)),
      'R2cv: %.3f'.ljust(15) % r2_score(y_train, ridge_cv.predict(x_train_scaled)),
      'R2: %.3f'.ljust(15) % r2_score(y_train, lasso.predict(x_train)),
      'R2cv: %.3f'.ljust(15) % r2_score(y_train, lasso_cv.predict(x_train_scaled)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_train, linear.predict(x_train)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_train, ridge.predict(x_train)),
      'MSEcv: %.3f'.ljust(15) % mean_squared_error(y_train, ridge_cv.predict(x_train_scaled)),
      'MSE: %.3f'.ljust(15) % mean_squared_error(y_train, lasso.predict(x_train)),
      'MSEcv: %.3f'.ljust(15) % mean_squared_error(y_train, lasso_cv.predict(x_train_scaled)))
print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, linear.predict(x_train)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, ridge.predict(x_train)),
      'MAEcv: %.3f'.ljust(15) % mean_absolute_error(y_train, ridge_cv.predict(x_train_scaled)),
      'MAE: %.3f'.ljust(15) % mean_absolute_error(y_train, lasso.predict(x_train)),
      'MAEcv: %.3f'.ljust(15) % mean_absolute_error(y_train, lasso_cv.predict(x_train_scaled)))
```

### Non lineare

```python
%%time

# Iperparametri scelti con grid search + k-fold cross validation
param_grid = {
    "n_neighbors": np.arange(start=10, stop=1000, step=10),
    "weights": ["distance", "uniform"],
    "metric": ["euclidean", "cityblock"],
}

grid_search = GridSearchCV(
    KNeighborsRegressor(),
    param_grid=param_grid,
    cv=KFold(n_splits=5, shuffle=True, random_state=0),
    n_jobs=-1,
    refit=True,
    verbose=1,
)

grid_search.fit(x_train_scaled, y_train)
knn_reg = grid_search.best_estimator_
```

```python
print(knn_reg.metric, knn_reg.n_neighbors, knn_reg.weights)
```

```python
%%time

# Ricerca di iperparametri come sopra ('None' a max_depth corrisponde a nessun limite di profondità)
param_grid = {
    'max_depth': list(np.arange(1, 102, step=20)), #+ [None],
    'min_samples_split': [2, 4, 8, 16, 32, 64, 128],
    'min_samples_leaf': [2, 4, 8, 16, 32, 64, 128],
    'criterion': ['squared_error']#, 'absolute_error' ]
}

grid_search = GridSearchCV(
    DecisionTreeRegressor(random_state=0),
    param_grid=param_grid,
    cv=KFold(n_splits=5, shuffle=True, random_state=0),
    n_jobs=-1,
    refit=True,
    verbose=1,
)

grid_search.fit(x_train, y_train)
dt_reg = grid_search.best_estimator_
```

```python
print(dt_reg.max_depth, dt_reg.min_samples_split, dt_reg.min_samples_leaf, dt_reg.criterion)
```

```python
dt_reg = DecisionTreeRegressor(max_depth=21, min_samples_split=2, min_samples_leaf=32, criterion='squared_error')
dt_reg.fit(x_train, y_train)

knn_reg = KNeighborsRegressor(metric='cityblock', weights='uniform', n_neighbors=50)
knn_reg.fit(x_train_scaled, y_train)
```

```python
print('Generalization error (TEST SET)\n')

print('DEC TREE'.ljust(15), 'KNN'.ljust(15))
# Nota: rimpiazzato y_ts con y_test
print('R2: %.3f'.ljust(14) % r2_score(y_test, dt_reg.predict(x_test)),
      'R2: %.3f'.ljust(14) % r2_score(y_test, knn_reg.predict(x_test_scaled)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_test, dt_reg.predict(x_test)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_test, knn_reg.predict(x_test_scaled)))

print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, dt_reg.predict(x_test)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_test, knn_reg.predict(x_test_scaled)))

print('\n' + '-'*30 + '\n')

print('Loss (TRAIN SET)\n')

print('DEC TREE'.ljust(15), 'KNN'.ljust(15))
# Nota: rimpiazzato y_tr con y_train
print('R2: %.3f'.ljust(14) % r2_score(y_train, dt_reg.predict(x_train)),
      'R2: %.3f'.ljust(14) % r2_score(y_train, knn_reg.predict(x_train_scaled)))

print('MSE: %.3f'.ljust(14) % mean_squared_error(y_train, dt_reg.predict(x_train)),
      'MSE: %.3f'.ljust(14) % mean_squared_error(y_train, knn_reg.predict(x_train_scaled)))

print('MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, dt_reg.predict(x_train)),
      'MAE: %.3f'.ljust(14) % mean_absolute_error(y_train, knn_reg.predict(x_train_scaled)))
```
